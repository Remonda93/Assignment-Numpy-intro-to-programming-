```python
import numpy  as np
```

```python
#q1)
array=np.array(np.random.randint(1,101,(3,3)),dtype=np.int16)
normalize_array=(array-array.min())/(array.max()-array.min())
print(normalize_array)
```

```python
#q2)
array=np.array([80,34,5,3,9,50,12,56,18,23])
#print(array)
num=int(input("Enter value"))
diff=abs(array-num)
#print(diff)
min_diff=np.min(diff) #the smallest diff

for i in range(len(array)):
    if(diff[i]==min_diff):
        print(array[i])
```

```python
#q3)
array=np.array(np.random.randint(1,101,(5,5)))
array_sorted=np.sort(array,axis=0) # orderd of columns
print(array_sorted)
print(array_sorted[-2]) #return row of before lasted
```

```python
#q4)
array=np.array(np.random.randint(1,101,(6,4)))
#print(array)
sum_array=np.sum(array,axis=1)
for i in range(len(sum_array)):
    if(sum_array[i] %2 ==0):
        print(i)

```

```python
#q5)
array=np.array(np.random.randint(1,50,30))
#print(array)
for i in range(len(array)):
    if ( array[i] %3==0 or array[i] % 5==0):
        array[i]=-1
print(array)        
```

```python
#q6)
array=np.array(np.random.randint(1,101,(3,3)))
#print(array)
average_arr=array.mean(axis=1)
#print(average_arr)
for i in range(len(average_arr)):
    array[i]=array[i]-average_arr[i]
print(array)    
```
